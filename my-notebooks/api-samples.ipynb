### Use generic function for all providers

Currently implemented:

- google or gemini
- groq
- local or lmstudio

Example with gemini:

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from utils.llm_client import get_llm_client

load_dotenv(override=True)

llm_client = get_llm_client(provider="google") # default is google
gemini_model = os.getenv("GEMINI_MODEL")

response = llm_client.chat.completions.create(model=gemini_model, messages=[{"role":"user", "content": "what is 2+2?"}])
print(response.choices[0].message.content)
```

## Google Gemini

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv(override=True)

GEMINI_BASE_URL = os.getenv("GEMINI_BASE_URL")
google_api_key = os.getenv("GOOGLE_API_KEY")
gemini_model = os.getenv("GEMINI_MODEL")
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)

response = gemini.chat.completions.create(model=gemini_model, messages=[{"role":"user", "content": "what is 2+2?"}])
print(response.choices[0].message.content)
```
