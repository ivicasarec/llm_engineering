To demonstrate your familiarity with OpenAI API, and also LM Studio, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from openai import OpenAI
from dotenv import load_dotenv
import os
from IPython.display import Markdown, display, update_display
```

```python
# constants
load_dotenv(override=True)

LMSTUDIO_BASE_URL = os.getenv("LMSTUDIO_BASE_URL")
lmstudio_api_key = os.getenv("LMSTUDIO_API_KEY")
lmstudio_model = os.getenv("LMSTUDIO_MODEL")

GEMINI_BASE_URL = os.getenv("GEMINI_BASE_URL")
google_api_key = os.getenv("GOOGLE_API_KEY")
gemini_model = os.getenv("GEMINI_MODEL")
```

```python
# set up environment
lmstudio = OpenAI(base_url=LMSTUDIO_BASE_URL, api_key=lmstudio_api_key)
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
system_prompt = "You are a helpful assistant for explaining code snippets."
```

```python
# Get gemini-2.5-flash to answer, with streaming
stream = gemini.chat.completions.create(
    model=gemini_model,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ],
    stream=True
)

response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Get google/gemma-3-4b to answer
response = lmstudio.chat.completions.create(
    model=lmstudio_model,
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ]
)
display(Markdown(response.choices[0].message.content))
```
