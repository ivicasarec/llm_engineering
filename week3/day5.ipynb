```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from utils.llm_client import get_llm_client

load_dotenv(override=True)

llm_client = get_llm_client(provider="google")
gemini_model = os.getenv("GEMINI_MODEL")
```

```python
# import importlib
# import visualizer
# importlib.reload(visualizer)

from visualizer import TokenPredictor, create_token_graph, visualize_predictions

message = "In one sentence, describe the color orange to someone who has never been able to see"
# model_name = "gpt-4.1-mini"

predictor = TokenPredictor(gemini_model, llm_client)
predictions = predictor.predict_tokens(message)
G = create_token_graph(gemini_model, predictions)
plt = visualize_predictions(G)
plt.show()
```

```python
print("Predictions:", predictions)
print("Number of tokens:", len(predictions))
if predictions:
    print("Generated text:", "".join([p["token"] for p in predictions]))
```
